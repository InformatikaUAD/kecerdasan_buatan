### Deskripsi Masalah:
Sebuah kampus memiliki **3 mata kuliah**, **3 dosen**, dan **3 slot waktu** (misalnya: Senin,
Selasa, Rabu). Setiap dosen hanya bisa mengajar di hari tertentu. Tujuannya adalah
membuat penjadwalan kuliah yang menghindari konflik jadwal (misal: dua mata
kuliah di slot waktu yang sama, atau dosen mengajar di hari yang tidak tersedia
baginya). Gunakan Hill Climbing untuk mencari solusi penjadwalan dengan jumlah
konflik paling sedikit.
Petunjuk:
Strategi Hill Climbing:
• State: Penjadwalan semua mata kuliah → {course: day}
• Neighbor: Mengubah hari salah satu mata kuliah ke hari lain
• Objective: Minimalkan konflik, yaitu:
a. Dosen mengajar di hari yang tidak tersedia
b. Dua kuliah terjadi di waktu yang sama

```python
import random
import copy

courses = ["Matematika", "Pemrograman", "Jaringan"]
lecturers = {
    "Matematika": "Dosen A",
    "Pemrograman": "Dosen B",
    "Jaringan": "Dosen C"
}
available_times = {
    "Dosen A": ["Senin", "Selasa"],
    "Dosen B": ["Selasa", "Rabu"],
    "Dosen C": ["Senin", "Rabu"]
}
timeslots = ["Senin", "Selasa", "Rabu"]

# Fungsi untuk menghasilkan penjadwalan acak
def random_schedule():
    return {course: random.choice(timeslots) for course in courses}

# Fungsi untuk menghitung konflik
def calculate_conflict(schedule):
    conflict = 0
    used_slots = {}
    for course, day in schedule.items():
        lecturer = lecturers[course]
        # Konflik: dosen tidak tersedia
        if day not in available_times[lecturer]:
            conflict += 1
        # Konflik: dua kuliah di waktu yang sama
        if day in used_slots:
            conflict += 1
        else:
            used_slots[day] = course
    return conflict

# Fungsi mencari tetangga
def get_neighbors(schedule):
    neighbors = []
    for course in courses:
        for day in timeslots:
            if schedule[course] != day:
                neighbor = copy.deepcopy(schedule)
                neighbor[course] = day
                neighbors.append(neighbor)
    return neighbors

# Hill Climbing Algorithm
def hill_climbing():
    current = random_schedule()
    current_conflict = calculate_conflict(current)
    
    while True:
        neighbors = get_neighbors(current)
        next_candidate = None
        next_conflict = current_conflict
        
        for neighbor in neighbors:
            conflict = calculate_conflict(neighbor)
            if conflict < next_conflict:
                next_candidate = neighbor
                next_conflict = conflict
        
        if next_conflict < current_conflict:
            current = next_candidate
            current_conflict = next_conflict
        else:
            break
    
    return current, current_conflict

# Eksekusi
solution, conflicts = hill_climbing()
print("Penjadwalan Optimal:")
for course, day in solution.items():
    print(f"- {course} dijadwalkan pada hari {day}")
print(f"Jumlah konflik: {conflicts}")
```

```python
import random
import copy

# 1. DATA DASAR
courses = ["Matematika", "Pemrograman", "Jaringan", "Statistika", "Basis Data"]
lecturers = {
    "Matematika": "Dosen A",
    "Pemrograman": "Dosen B",
    "Jaringan": "Dosen C",
    "Statistika": "Dosen D",
    "Basis Data": "Dosen E"
}
available_times = {
    "Dosen A": ["Senin", "Selasa"],
    "Dosen B": ["Selasa", "Rabu"],
    "Dosen C": ["Senin", "Rabu"],
    "Dosen D": ["Senin", "Kamis"],
    "Dosen E": ["Selasa", "Jumat"]
}
timeslots = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat"]
rooms = ["R1", "R2", "R3"]

# 2. FUNGSI MEMBUAT JADWAL ACAK
def random_schedule():
    return {course: (random.choice(timeslots), random.choice(rooms)) for course in courses}

# 3. FUNGSI MENGHITUNG KONFLIK
def calculate_conflict(schedule):
    conflict = 0
    used_slots = {}  # key: (hari, ruang), value: list of course

    for course, (day, room) in schedule.items():
        lecturer = lecturers[course]

        # Konflik: dosen tidak tersedia
        if day not in available_times[lecturer]:
            conflict += 1

        # Konflik: ruangan sudah dipakai di hari yang sama
        if (day, room) in used_slots:
            conflict += 1
        else:
            used_slots[(day, room)] = course

    return conflict

# 4. FUNGSI MENCARI TETANGGA
def get_neighbors(schedule):
    neighbors = []
    for course in courses:
        current_day, current_room = schedule[course]
        for day in timeslots:
            for room in rooms:
                if (day, room) != (current_day, current_room):
                    neighbor = copy.deepcopy(schedule)
                    neighbor[course] = (day, room)
                    neighbors.append(neighbor)
    return neighbors

# 5. HILL CLIMBING
def hill_climbing():
    current = random_schedule()
    current_conflict = calculate_conflict(current)

    while True:
        neighbors = get_neighbors(current)
        next_candidate = None
        next_conflict = current_conflict

        for neighbor in neighbors:
            conflict = calculate_conflict(neighbor)
            if conflict < next_conflict:
                next_candidate = neighbor
                next_conflict = conflict

        if next_conflict < current_conflict:
            current = next_candidate
            current_conflict = next_conflict
        else:
            break

    return current, current_conflict

# 6. HILL CLIMBING DENGAN RESTART
def hill_climbing_with_restart(restarts=30):
    best_solution = None
    best_conflict = float('inf')

    for _ in range(restarts):
        solution, conflict = hill_climbing()
        if conflict < best_conflict:
            best_solution = solution
            best_conflict = conflict
        if best_conflict == 0:
            break

    return best_solution, best_conflict

# 7. JALANKAN PROGRAM
final_schedule, total_conflict = hill_climbing_with_restart()

# 8. TAMPILKAN HASIL
print("Penjadwalan Optimal:")
for course, (day, room) in final_schedule.items():
    print(f"- {course} dijadwalkan pada hari {day} di ruang {room}")
print(f"\nJumlah konflik: {total_conflict}")
```
